```python
from ultralytics import YOLO
import os
import glob
import matplotlib.pyplot as plt
from PIL import Image
```

```python
MODEL_PATH = r"C:\Users\User\Downloads\Textile_fabrics\best.pt"
DATA_YAML = r"C:\Users\User\Downloads\Textile_fabrics\final project dataset\data.yaml"
DATASET_DIR = r"C:\Users\User\Downloads\Textile_fabrics\final project dataset"
TEST_IMAGES = r"C:\Users\User\Downloads\Textile_fabrics\final project dataset\test\images" # or your test path
```

```python
print("Checking files...\n")

print("Model:")
print(MODEL_PATH)
print("Exists:", os.path.exists(MODEL_PATH))

print("\nDataset YAML:")
print(DATA_YAML)
print("Exists:", os.path.exists(DATA_YAML))

print("\nDataset directory:")
print(DATASET_DIR)
print("Exists:", os.path.exists(DATASET_DIR))

print("\nTest images:")
print(TEST_IMAGES)
print("Exists:", os.path.exists(TEST_IMAGES))
```

```python
print("Dataset contents:\n")

for item in os.listdir(DATASET_DIR):
    print(item)
```

```python
test_images = glob.glob(
    os.path.join(TEST_IMAGES, "*")
)

print(f"\nNumber of test images: {len(test_images)}")
```

```python
model = YOLO(MODEL_PATH)

print("YOLOv8 model loaded successfully.")
```

```python
val_results = model.val(
    data=DATA_YAML,
    split="val",
    imgsz=640,
    batch=16,
    device="cpu",
    plots=True,
    verbose=True
)
```

```python
print("=" * 60)
print("VALIDATION RESULTS")
print("=" * 60)

print(f"Precision : {val_results.box.mp:.4f}")
print(f"Recall    : {val_results.box.mr:.4f}")
print(f"mAP50     : {val_results.box.map50:.4f}")
print(f"mAP50-95  : {val_results.box.map:.4f}")
```

```python
print("=" * 80)
print("PER-CLASS VALIDATION PERFORMANCE")
print("=" * 80)

print(
    f"{'ID':<5}"
    f"{'Class':<20}"
    f"{'Precision':<12}"
    f"{'Recall':<12}"
    f"{'mAP50':<12}"
    f"{'mAP50-95':<12}"
)

print("-" * 80)

for i, class_id in enumerate(val_results.box.ap_class_index):

    class_id = int(class_id)
    class_name = model.names[class_id]

    precision = val_results.box.p[i]
    recall = val_results.box.r[i]
    ap50 = val_results.box.ap50[i]
    ap = val_results.box.ap[i]

    print(
        f"{class_id:<5}"
        f"{class_name:<20}"
        f"{precision:<12.4f}"
        f"{recall:<12.4f}"
        f"{ap50:<12.4f}"
        f"{ap:<12.4f}"
    )
```

```python
test_results = model.val(
    data=DATA_YAML,
    split="test",
    imgsz=640,
    batch=16,
    device="cpu",
    plots=True,
    verbose=True
)
```

```python
print("=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)

print(f"Precision : {test_results.box.mp:.4f}")
print(f"Recall    : {test_results.box.mr:.4f}")
print(f"mAP50     : {test_results.box.map50:.4f}")
print(f"mAP50-95  : {test_results.box.map:.4f}")
```

```python
print("=" * 80)
print("PER-CLASS TEST PERFORMANCE")
print("=" * 80)

print(
    f"{'ID':<5}"
    f"{'Class':<20}"
    f"{'Precision':<12}"
    f"{'Recall':<12}"
    f"{'mAP50':<12}"
    f"{'mAP50-95':<12}"
)

print("-" * 80)

for i, class_id in enumerate(test_results.box.ap_class_index):

    class_id = int(class_id)
    class_name = model.names[class_id]

    precision = test_results.box.p[i]
    recall = test_results.box.r[i]
    ap50 = test_results.box.ap50[i]
    ap = test_results.box.ap[i]

    print(
        f"{class_id:<5}"
        f"{class_name:<20}"
        f"{precision:<12.4f}"
        f"{recall:<12.4f}"
        f"{ap50:<12.4f}"
        f"{ap:<12.4f}"
    )
    
```

```python
prediction_results = model.predict(
    source=TEST_IMAGES,
    imgsz=640,
    conf=0.25,
    save=True,
    save_txt=True,
    save_conf=True,
    device="cpu",
    verbose=True
)
```

```python
prediction_folder = prediction_results[0].save_dir

prediction_images = glob.glob(
    os.path.join(str(prediction_folder), "*")
)

print("Prediction folder:")
print(prediction_folder)

print(f"\nNumber of prediction files: {len(prediction_images)}")
```
